## Contagem regressiva

Um evento especial está programado para começar em 10 segundos. Crie uma contagem regressiva que começa em 10 e vai até 0, com uma pausa de um segundo entre cada número.


```python
import time

contagem_regressiva = 10

while contagem_regressiva > 0:
    print(contagem_regressiva)
    time.sleep(1)
    contagem_regressiva -= 1

print("Feliz Ano Novo!")
```

## Formatação de tempo

Uma empresa quer exibir a data e a hora atual em seu site no formato "Dia da semana, dia do mês de mês de ano, horas:minutos". Crie um script Python que mostra a data e a hora atuais neste formato.



```python
import locale
import time

# Definir a localização para português.
locale.setlocale(locale.LC_TIME, 'pt_BR.UTF-8')

hora_atual = time.localtime()
print(hora_atual)
hora_formatada = time.strftime("%H:%M", hora_atual)
dia_semana = time.strftime("%A", hora_atual)
dia = time.strftime("%d", hora_atual)
mes = time.strftime("%B", hora_atual)
ano = time.strftime("%Y", hora_atual)

print(f"{dia_semana}, {dia} de {mes} de {ano}, {hora_formatada}.")
```
