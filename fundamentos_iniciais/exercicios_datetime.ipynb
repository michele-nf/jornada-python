## Oferecendo desconto para cliente com base na última compra

Suponha que você está trabalhando para uma empresa que deseja rastrear a atividade do cliente. Uma métrica que eles estão interessados é o tempo que passou desde a última transação do cliente. Se for muito tempo, eles podem oferecer um desconto para o cliente. Crie um script Python que mostra quanto tempo se passou desde a última compra do cliente. Se faz mais de 30 dias, mostre uma mensagem oferecendo um desconto para o cliente.

```python
from datetime import datetime

tempo_ultima_transacao = input('Digite a data e hora da última transação (formato: DD/MM/AAAA HH:MM:SS): ')
tempo_ultima_transacao = datetime.strptime(tempo_ultima_transacao, '%d/%m/%Y %H:%M:%S').timestamp()
tempo_atual = datetime.now().timestamp()

tempo_decorrido_segundos = tempo_atual - tempo_ultima_transacao
tempo_decorrido_dias = tempo_decorrido_segundos / 86400  # Convertendo segundos para dias

print(f'Última transação foi há {int(tempo_decorrido_dias)} dias.')

if tempo_decorrido_segundos >= 2592000:
    print('Parabéns! Você ganhou um desconto.')
```

## Data e hora em diferentes fusos horários

Uma empresa tem escritórios em São Paulo, Nova York e Tóquio. Crie um script Python que mostra a data e hora atuais nesses três fusos horários. Exiba, também, se estes escritórios estão abertos ou fechados (9h às 17h).


```python
from datetime import datetime
from zoneinfo import ZoneInfo

def verificar_horario_funcionamento(horario_cidade):
    if 9 <= horario_cidade.hour < 17:
        return "aberto"
    else:
        return "fechado"

fuso_horario_sao_paulo = ZoneInfo('America/Sao_Paulo')
agora_sao_paulo = datetime.now(fuso_horario_sao_paulo)
print("Data e hora em São Paulo:", agora_sao_paulo.strftime('%d/%m/%Y %H:%M:%S'))
print(f"O escritório em São Paulo está {verificar_horario_funcionamento(agora_sao_paulo)}.")

fuso_horario_nova_york = ZoneInfo('America/New_York')
agora_nova_york = datetime.now(fuso_horario_nova_york)
print("Data e hora em Nova York:", agora_nova_york.strftime('%d/%m/%Y %H:%M:%S'))
print(f"O escritório em Nova York está {verificar_horario_funcionamento(agora_nova_york)}.")

fuso_horario_toquio = ZoneInfo('Asia/Tokyo')
agora_toquio = datetime.now(fuso_horario_toquio)
print("Data e hora em Tóquio:", agora_toquio.strftime('%d/%m/%Y %H:%M:%S'))   
print(f"O escritório em Tóquio está {verificar_horario_funcionamento(agora_toquio)}.")
```
