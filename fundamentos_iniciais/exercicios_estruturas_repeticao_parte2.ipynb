#### 1. Faça um programa que peça uma nota, entre zero e dez. Mostre uma mensagem caso o valor seja inválido e continue pedindo até que o usuário informe um valor válido.

````python
while True:
        nota = float(input('Insira uma nota de zero a dez: '))
        if 0 <= nota <= 10:
            print(f'Nota informada: {nota}')
            break
        else:
            print("Nota inválida. Tente novamente.")

````

#### 2. Faça um programa que leia um nome de usuário e a sua senha e não aceite a senha igual ao nome do usuário, mostrando uma mensagem de erro e voltando a pedir as informações.

````python
while True:
    nome = input('Insira o nome de usuário: ')
    senha = input('Insira a senha: ')
    if nome != senha:
        print(f'Usuário: {nome}, Senha: {senha}')
        break
    else:
        print("Usuário e senha não podem ser iguais. Tente novamente.")
````

#### 3. Faça um programa que leia e valide as seguintes informações (e para cada uma delas, continue pedindo a informação até o usuário inserir corretamente):
##### Nome: maior que 3 caracteres;
##### Idade: entre 0 e 150;
##### Salário: maior que zero;
##### Sexo: 'f' ou 'm';
##### Estado Civil: 's', 'c', 'v', 'd';

````python
while True:
    nome = input('Insira o nome: ')
    tamanho_nome = len(nome)
    if tamanho_nome > 3:
        print(f'Nome: {nome}')
        break
    else:
        print("O nome deve ter mais de 3 caracteres. Tente novamente.")

while True:
    idade = int(input('Insira a idade: '))
    if 0 <= idade <= 150:
        print(f'Idade: {idade}')
        break
    else:
        print("Idade inválida. Deve ser entre 0 e 150. Tente novamente.")

while True:
    salario = float(input('Insira o salário: '))
    if salario > 0:
        print(f'Salário: R$ {salario}')
        break
    else:
        print("Salário inválido. Deve ser maior que zero. Tente novamente.")

while True:
    sexo = input('Insira o sexo (M/F): ').upper()
    if sexo in ['M', 'F']:
        print(f'Sexo: {sexo}')
        break
    else:
        print("Sexo inválido. Deve ser 'M' ou 'F'. Tente novamente.")

while True:
    estado_civil = input('Insira o estado civil (S, C, V, D): ').upper()
    if estado_civil in ['S', 'C', 'V', 'D']:
        print(f'Estado Civil: {estado_civil}')
        break
    else:
        print("Estado civil inválido. Deve ser 'S', 'C', 'V' ou 'D'. Tente novamente.")
````

#### 4. Supondo que a população de um país A seja da ordem de 80000 habitantes com uma taxa anual de crescimento de 3% e que a população de B seja 200000 habitantes com uma taxa de crescimento de 1.5%. Faça um programa que calcule e escreva o número de anos necessários para que a população do país A ultrapasse ou iguale a população do país B, mantidas as taxas de crescimento.

````python
populacao_paisA = 80000
populacao_paisB = 200000
taxa_crescimento_A = 0.03
taxa_crescimento_B = 0.015
contador_anos = 0

while populacao_paisA < populacao_paisB:
    populacao_paisA += int(populacao_paisA * taxa_crescimento_A)
    populacao_paisB += int(populacao_paisB * taxa_crescimento_B)
    contador_anos += 1

print(f'População País A: {populacao_paisA}, População País B: {populacao_paisB}, Anos: {contador_anos} anos')
````

#### 6. Faça um programa que peça para o usuário inserir o faturamento dos últimos 5 meses (individualmente) e informe o maior faturamento

````python
faturamento_mes = []
contador_mes = 1

while len(faturamento_mes) < 5:
    faturamento = float(input(f'Informe o faturamento do mês {contador_mes}: '))
    faturamento_mes.append(faturamento)
    contador_mes += 1

maior_faturamento_mes = max(faturamento_mes)

print(f'O maior faturamento do período foi: R$ {maior_faturamento_mes:.2f}')



````

#### 7. Faça um programa que peça para o usuário inserir o faturamento dos últimos 5 meses (individualmente) e informe o faturamento total (soma) e o faturamento médio por mês (média).

````python
faturamento_mes = []
contador_mes = 1

while len(faturamento_mes) < 5:
    faturamento = float(input(f'Informe o faturamento do mês {contador_mes}: '))
    faturamento_mes.append(faturamento)
    contador_mes += 1

faturamento_total = sum(faturamento_mes)
faturamento_medio = faturamento_total / len(faturamento_mes)
print(f'O faturamento total do período foi: R$ {faturamento_total:.2f}')
print(f'O faturamento médio do período foi: R$ {faturamento_medio:.2f}')
````

#### 8. Faça um programa que consiga categorizar a idade das equipes de uma empresa. Faça um programa que peça para n pessoas a sua idade, ao final o programa devera verificar se a média de idade da equipe varia entre 0 e 25 (jovem) ,26 e 60 (sênior) e maior que 60 (idosa); e então, dizer se a equipe é jovem, sênior ou idosa, conforme a média calculada.

````python
idade = 0
idades = []

while idade != '':
    idade = input('Informe a idade (ou pressione Enter para sair): ')
    if idade == '':
        break
    idades.append(int(idade))
    
media_idade = sum(idades) / len(idades)

if 0 <= media_idade <= 25:
    print('A equipe é jovem.')
elif 25 < media_idade <= 60:
    print('A equipe é sênior.')
elif media_idade > 60:
    print('A equipe é idosa.')
print(f'A média de idade da equipe é: {media_idade:.2f}')
    
    
````

#### 9. Numa eleição existem três candidatos. Faça um programa que peça o número total de eleitores. Peça para cada eleitor votar e ao final mostrar o número de votos de cada candidato.

````python
eleitores = int(input('Informe o total de eleitores: '))
candidato_A = 0
candidato_B = 0
candidato_C = 0

for eleito in range(eleitores):
    voto = input('Digite seu voto (1 para Candidato A, 2 para Candidato B, 3 para Candidato C): ')
    if voto == '1':
        candidato_A += 1
    elif voto == '2':
        candidato_B += 1
    elif voto == '3':
        candidato_C += 1

print(f'Total de votos para Candidato A: {candidato_A}')
print(f'Total de votos para Candidato B: {candidato_B}')
print(f'Total de votos para Candidato C: {candidato_C}')
````

#### 10. Faça um programa que calcule o valor total investido por um colecionador em sua coleção de CDs e o valor médio gasto em cada um deles. O usuário deverá informar a quantidade de CDs e o valor para em cada um.

````python
cds = int(input('Informe o número de CDs: '))
contador_cds = 1
valor_cds = []

for cd in range(cds):
    valor_cd = float(input(f'Informe o valor do CD {contador_cds}: '))
    contador_cds += 1
    valor_cds.append(valor_cd)
    print(f'Valor do CD {contador_cds - 1}: R$ {valor_cd:.2f}')

media_valor_cds = sum(valor_cds) / len(valor_cds)
print(f'O valor médio dos CDs é: R$ {media_valor_cds:.2f}')

````
