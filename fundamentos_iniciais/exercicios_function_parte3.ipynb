## 1. Cálculo do Percentual e da Lista de Vendedores

- Queremos criar uma function que consiga identificar os vendedores que bateram uma meta, mas além disso, consigo já me dar como resposta o cálculo do % da lista de vendedores que bateu a meta (para eu não precisar calcular manualmente depois)
- Essa function deve receber 2 informações como parâmetro: a meta e um dicionário com os vendedores e suas vendas. E me dar 2 respostas: uma lista com o nome dos vendedores que bateram a meta e o % de vendedores que bateu a meta.

```python
meta = 10000
vendas = {
    'João': 15000,
    'Julia': 27000,
    'Marcus': 9900,
    'Maria': 3750,
    'Ana': 10300,
    'Alon': 7870,
}

```

```python
#crie sua function aqui
def verificar_vendedor_atingiu_meta(meta, vendas):
    lista_vendedores_meta = []
    for venda in vendas:
        vendedor, valor = venda, vendas[venda]
        if valor >= meta:
            lista_vendedores_meta.append(vendedor)
    
    tamanho_lista_vendedores = len(vendas)
    tamanho_lista_vendedores_meta = len(lista_vendedores_meta)
    percentagem_vendedores_meta = (tamanho_lista_vendedores_meta / tamanho_lista_vendedores) * 100
    
    return lista_vendedores_meta, percentagem_vendedores_meta
```

```python
#aplique sua function nas informações para ver se está funcionando
print(f"Percentagem de vendedores que atingiram a meta: {verificar_vendedor_atingiu_meta(meta, vendas)[1]:.2f}%")
print(f"Vendedores que atingiram a meta: {verificar_vendedor_atingiu_meta(meta, vendas)[0]}")
```
