## 1. Faturamento do Melhor e do Pior Mês do Ano

Qual foi o valor de vendas do melhor mês do Ano?
E valor do pior mês do ano?

```python
meses = ['jan', 'fev', 'mar', 'abr', 'mai', 'jun', 'jul', 'ago', 'set', 'out', 'nov', 'dez']
vendas_1sem = [25000, 29000, 22200, 17750, 15870, 19900]
vendas_2sem = [19850, 20120, 17540, 15555, 49051, 9650]

vendas_ano = vendas_1sem + vendas_2sem

maior_valor = max(vendas_ano)
menor_valor = min(vendas_ano)

indice_maior_valor = vendas_ano.index(maior_valor)
indice_menor_valor = vendas_ano.index(menor_valor)

mes_maior_valor = meses[indice_maior_valor] 
mes_menor_valor = meses[indice_menor_valor]

print(f"O melhor mês do ano de vendas foi {mes_maior_valor} e o pior mês do ano de vendas foi {mes_menor_valor}. ")

```

## 2. Continuação

Agora relacione as duas listas para printar 'O melhor mês do ano foi {} com {} vendas' e o mesmo para o pior mês do ano.

Calcule também o faturamento total do Ano e quanto que o melhor mês representou do faturamento total.

Obs: Para o faturamento total, pode usar a função sum(lista) que soma todos os itens de uma lista

```python
print(f"O melhor mês do ano foi {mes_maior_valor} com {maior_valor} vendas e o pior mês do ano foi {mes_menor_valor} com {menor_valor} vendas. ")

faturamento_ano = sum(vendas_ano)
representacao_melhor_mes = (maior_valor * 100) / faturamento_ano

print(f"O faturamento total do ano foi de {faturamento_ano} reais e o melhor mês representou {representacao_melhor_mes}% do faturamento.")
```
