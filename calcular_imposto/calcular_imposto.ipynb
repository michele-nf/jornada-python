### 1- Imposto a pagar no Lucro Presumido

- 5% sobre faturamento de ISS (mensal)
- 0,65% de PIS sobre faturamento, (mensal)
- 3% de COFINS sobre faturmaneto, (mensal)
- 4.8% de IR (trimestral)
- 10% de IR Adicional sobre o que ultrapassar 20mil do faturamento (trimestral)
- CSLL: 2,88% sobre faturamento (trimestral)

```python
faturamento = {
    'jan': 'R$ 95.141,98',
    'fev': 'R$ 95.425,16',
    'mar': 'R$ 89.716,31',
    'abr': 'R$ 78.459,99',
    'mai': 'R$ 71.087,28',
    'jun': 'R$ 83.911,06',
    'jul': 'R$ 56.467,26',
    'ago': 'R$ 88.513,58',
    'set': 'R$ 66.552,49',
    'out': 'R$ 80.164,07',
    'nov': 'R$ 66.964,33',
    'dez': 'R$ 71.525,25',
}

### 1- Objetivo: Imposto a pagar no Lucro Presumido

# Premissas:
# - 5% sobre faturamento de ISS (mensal)
# - 0,65% de PIS sobre faturamento, (mensal)
# - 3% de COFINS sobre faturmaneto, (mensal)
# - 4.8% de IR (trimestral)
# - 10% de IR Adicional sobre o que ultrapassar 20mil do faturamento (trimestral)
# - CSLL: 2,88% sobre faturamento (trimestral)

# você precisa inserir no sistema um dicionário no formato:
# resultado = {
#     'mes': (faturamento, imposto_mensal, imposto_trimestral),
# }

#eliminar 'R$' e converter para float

resultado = {}  

for mes in faturamento:
    faturamento[mes] = float(faturamento[mes].replace('R$ ', '').replace('.', '').replace(',', '.'))

print(faturamento)

def imposto_mensal(faturamento):
    iss = faturamento * 0.05
    pis = faturamento * 0.0065
    cofins = faturamento * 0.03
    return iss + pis + cofins

def imposto_trimestral(faturamento_trimestre):
    ir = faturamento_trimestre * 0.048
    ir_adicional = 0
    if faturamento_trimestre > 20000:
        ir_adicional = (faturamento_trimestre - 20000) * 0.10
    csll = faturamento_trimestre * 0.0288
    return ir + ir_adicional + csll

meses = list(faturamento.keys())
print(meses)
for mes in meses:
    faturamento_mes = faturamento[mes]
    imposto_mensal_valor = imposto_mensal(faturamento_mes)
    imposto_trimestral_valor = imposto_trimestral(faturamento_mes)
    resultado[mes] = (faturamento_mes, imposto_mensal_valor, imposto_trimestral_valor)

print(resultado)
    
```
